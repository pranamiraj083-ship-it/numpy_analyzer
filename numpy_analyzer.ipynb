```python
import numpy as np


def indexing_slicing(arr):

    while True:

        print("\nChoose an operation:")
        print("1. Indexing")
        print("2. Slicing")
        print("3. Go Back")

        choice = int(input("Enter your choice: "))

        if choice == 1:

            if arr.ndim == 1:

                index = int(input("\nEnter the index: "))

                print("\nElement:")
                print(arr[index])

            elif arr.ndim == 2:

                row = int(input("\nEnter the row index: "))
                col = int(input("Enter the column index: "))

                print("\nElement:")
                print(arr[row, col])

            elif arr.ndim == 3:

                x = int(input("\nEnter dimension 1 index: "))
                y = int(input("Enter dimension 2 index: "))
                z = int(input("Enter dimension 3 index: "))

                print("\nElement:")
                print(arr[x, y, z])

        elif choice == 2:

            if arr.ndim == 1:

                start = int(input("\nEnter the start index: "))
                end = int(input("Enter the end index: "))

                print("\nSliced Array:")
                print(arr[start:end])

            elif arr.ndim == 2:

                row_range = input(
                    "\nEnter the row range (start:end): "
                )

                col_range = input(
                    "Enter the column range (start:end): "
                )

                row_start, row_end = map(
                    int, row_range.split(":")
                )

                col_start, col_end = map(
                    int, col_range.split(":")
                )

                print("\nSliced Array:")
                print(arr[row_start:row_end, col_start:col_end])

            elif arr.ndim == 3:

                x_range = input(
                    "\nEnter dimension 1 range (start:end): "
                )

                y_range = input(
                    "Enter dimension 2 range (start:end): "
                )

                z_range = input(
                    "Enter dimension 3 range (start:end): "
                )

                x_start, x_end = map(
                    int, x_range.split(":")
                )

                y_start, y_end = map(
                    int, y_range.split(":")
                )

                z_start, z_end = map(
                    int, z_range.split(":")
                )

                print("\nSliced Array:")
                print(
                    arr[
                        x_start:x_end,
                        y_start:y_end,
                        z_start:z_end
                    ]
                )

        elif choice == 3:
            break

        else:
            print("Invalid choice.")


def create_array():

    print("\nArray Creation:")
    print("Select the type of array to create:")
    print("1. 1D Array")
    print("2. 2D Array")
    print("3. 3D Array")

    choice = int(input("Enter your choice: "))

    if choice == 1:

        n = int(input("\nEnter the number of elements: "))

        elements = list(map(int, input(
            f"Enter {n} elements separated by space: "
        ).split()))

        arr = np.array(elements)

    elif choice == 2:

        rows = int(input("\nEnter the number of rows: "))
        cols = int(input("Enter the number of columns: "))

        elements = list(map(int, input(
            f"Enter {rows * cols} elements separated by space: "
        ).split()))

        arr = np.array(elements).reshape(rows, cols)

    elif choice == 3:

        d1 = int(input("\nEnter the size of dimension 1: "))
        d2 = int(input("Enter the size of dimension 2: "))
        d3 = int(input("Enter the size of dimension 3: "))

        elements = list(map(int, input(
            f"Enter {d1 * d2 * d3} elements separated by space: "
        ).split()))

        arr = np.array(elements).reshape(d1, d2, d3)

    else:

        print("Invalid choice.")
        return None

    print("\nArray created successfully:")
    print(arr)

    indexing_slicing(arr)

    return arr


def mathematical_operations(arr):

    print("\nMathematical Operations:")
    print("Choose a mathematical operation:")
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")

    choice = int(input("Enter your choice: "))

    size = arr.size

    elements = list(map(int, input(
        f"\nEnter the same-size array elements "
        f"({size} elements separated by space): "
    ).split()))

    second_arr = np.array(elements).reshape(arr.shape)

    print("\nOriginal Array:")
    print(arr)

    print("\nSecond Array:")
    print(second_arr)

    if choice == 1:

        result = arr + second_arr

        print("\nResult of Addition:")
        print(result)

    elif choice == 2:

        result = arr - second_arr

        print("\nResult of Subtraction:")
        print(result)

    elif choice == 3:

        result = arr * second_arr

        print("\nResult of Multiplication:")
        print(result)

    elif choice == 4:

        result = arr / second_arr

        print("\nResult of Division:")
        print(result)

    else:

        print("Invalid choice.")


def combine_split(arr):

    print("\nCombine or Split Arrays:")
    print("Choose an option:")
    print("1. Combine Arrays")
    print("2. Split Array")

    choice = int(input("Enter your choice: "))

    if choice == 1:

        size = arr.size

        elements = list(map(int, input(
            f"\nEnter the elements of another array to combine "
            f"({size} elements separated by space): "
        ).split()))

        second_arr = np.array(elements).reshape(arr.shape)

        print("\nOriginal Array:")
        print(arr)

        print("\nSecond Array:")
        print(second_arr)

        if arr.ndim == 2:

            result = np.vstack((arr, second_arr))

            print("\nCombined Array (Vertical Stack):")
            print(result)

        else:

            result = np.concatenate((arr, second_arr))

            print("\nCombined Array:")
            print(result)

    elif choice == 2:

        parts = int(input("\nEnter the number of parts to split: "))

        result = np.array_split(arr, parts)

        print("\nSplit Array:")

        for i, part in enumerate(result, 1):

            print(f"\nPart {i}:")
            print(part)

    else:

        print("Invalid choice.")


def search_sort_filter(arr):

    print("\nSearch, Sort, and Filter:")
    print("Choose an option:")
    print("1. Search a value")
    print("2. Sort the array")
    print("3. Filter values")

    choice = int(input("Enter your choice: "))

    print("\nOriginal Array:")
    print(arr)

    if choice == 1:

        value = int(input("\nEnter the value to search: "))

        positions = np.where(arr == value)

        if len(positions[0]) > 0:

            print("\nValue found at position:")
            print(positions)

        else:

            print("\nValue not found.")

    elif choice == 2:

        if arr.ndim == 2:

            result = np.sort(arr, axis=1)

        else:

            result = np.sort(arr)

        print("\nSorted Array:")
        print(result)

        if arr.ndim == 2:
            print("(Sorting applied row-wise.)")

    elif choice == 3:

        value = int(input(
            "\nEnter a value to filter values greater than it: "
        ))

        result = arr[arr > value]

        print("\nFiltered Values:")
        print(result)

    else:

        print("Invalid choice.")


def aggregates_statistics(arr):

    print("\nAggregates and Statistics:")
    print("Choose an aggregate/statistical operation:")
    print("1. Sum")
    print("2. Mean")
    print("3. Median")
    print("4. Standard Deviation")
    print("5. Variance")

    choice = int(input("Enter your choice: "))

    print("\nOriginal Array:")
    print(arr)

    if choice == 1:

        print("\nSum of Array:", np.sum(arr))

    elif choice == 2:

        print("\nMean of Array:", np.mean(arr))

    elif choice == 3:

        print("\nMedian of Array:", np.median(arr))

    elif choice == 4:

        print("\nStandard Deviation of Array:", np.std(arr))

    elif choice == 5:

        print("\nVariance of Array:", np.var(arr))

    else:

        print("Invalid choice.")


def main():

    arr = None

    print("Welcome Start:")
    print("Welcome to the NumPy Analyzer!")
    print("======================================")

    while True:

        print("\nChoose an option:")
        print("1. Create a NumPy Array")
        print("2. Perform Mathematical Operations")
        print("3. Combine or Split Arrays")
        print("4. Search, Sort, or Filter Arrays")
        print("5. Compute Aggregates and Statistics")
        print("6. Exit")

        choice = int(input("Enter your choice: "))

        if choice == 1:

            arr = create_array()

        elif choice == 2:

            if arr is None:
                print("\nPlease create an array first.")
            else:
                mathematical_operations(arr)

        elif choice == 3:

            if arr is None:
                print("\nPlease create an array first.")
            else:
                combine_split(arr)

        elif choice == 4:

            if arr is None:
                print("\nPlease create an array first.")
            else:
                search_sort_filter(arr)

        elif choice == 5:

            if arr is None:
                print("\nPlease create an array first.")
            else:
                aggregates_statistics(arr)

        elif choice == 6:

            print("\nThank you for using the NumPy Analyzer! Goodbye!")
            break

        else:

            print("\nInvalid choice. Please try again.")

main()
```
